```python
import yt_dlp, os
print("yt_dlp import OK")
```

```python
os.makedirs("videos", exist_ok=True)
```

```python
url = "https://www.youtube.com/watch?v=ig49C04bJt0&t=55s"
```

```python
ydl_opts = {
    # Progressive MP4 bevorzugen, sonst bestes MP4-Fallback
    "format": "best[ext=mp4][vcodec!=av01]/best[ext=mp4]/best",
    "outtmpl": "videos/%(title).200B.%(ext)s",
    "noprogress": True,
}
```

```python
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([url])
```
